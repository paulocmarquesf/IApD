## Exemplo mínimo verificável

Mensagem: “Meu pedido P0531 está atrasado; quero a previsão.” Campos esperados: category=delivery, order_id=P0531, issue=delay. A mensagem não informa uma data de entrega confirmada: delivery_date deve permanecer ausente ou nulo.

Uma saída com delivery_date="amanhã" pode ser JSON válido, mas não é semanticamente justificada. A validade estrutural e a validade factual são propriedades diferentes.

```python
import re, json
message = "Meu pedido P0531 está atrasado; preciso de ajuda."
order = re.search(r"P\d{4}", message)
result = {"category": "delivery" if "atrasado" in message else "unknown", "order_id": order.group() if order else None}
print(json.dumps(result, ensure_ascii=False, indent=2))
```
