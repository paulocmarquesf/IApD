## Abrir os dados locais

A localização é identificada automaticamente. A demonstração utiliza arquivos fornecidos com o curso.

```python
from pathlib import Path
folder = Path.cwd()
for base in [folder, *folder.parents]:
    if (base / "dados").is_dir():
        DATA = base / "dados"
        break
else:
    raise FileNotFoundError("Abra o notebook a partir do repositório completo, que contém a pasta dados/")
print("Pasta de dados:", DATA)
```

## Exemplo mínimo verificável

O mapa aponta Brava como concorrente mais próxima de Atlas. A participação de Brava passa de 19% a 22%. Logo, aumentou três pontos percentuais. Se só a participação de 2026 foi consultada, afirmar a variação sem buscar 2025 é injustificado.

Três pontos percentuais ≠ crescimento relativo de 3%. Crescimento relativo = (22−19)/19≈15,8%. A interface deve distinguir ambos.

```python
import csv
with open(DATA / "brands_market.csv", encoding="utf-8-sig") as f:
    brands = {row["brand"]: row for row in csv.DictReader(f)}
brava = brands["Brava"]
share_change = float(brava["share_2026"])-float(brava["share_2025"])
print("Brava:",share_change,"pontos percentuais")
print("Crescimento relativo:",share_change/float(brava["share_2025"]))
```
