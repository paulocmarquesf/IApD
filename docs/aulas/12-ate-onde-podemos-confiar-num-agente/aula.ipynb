## Exemplo mínimo verificável

Probabilidade independente de sucesso p=0,90. Em três tentativas: pass@3=1−(1−0,90)^3=0,999. Êxito em todas as três: pass^3=0,90^3=0,729. São critérios diferentes e não substituem teste real.

Uma trajetória consulta → verifica → executa reembolso sem aprovação. Mesmo que o valor seja correto, ela viola uma condição de segurança. A ferramenta deve impedir a ação.

```python
import json
p,k=0.90,3
print("pass@3:",1-(1-p)**k)
print("pass^3:",p**k)
traces=json.loads((DATA/"agent_traces.json").read_text(encoding="utf-8"))
for x in traces:
    print(x["case"],"aprovado:",x["approved"],"trajetória válida:",x["valid"])
```
