## Exemplo mínimo verificável

Considere u=(1,0), v=(0,8;0,6) e w=(0,1). Todos têm norma 1. Portanto cos(u,v)=0,8 e cos(u,w)=0. A primeira dupla é mais alinhada. Isso é uma propriedade geométrica, não prova que seus autores são consumidores substituíveis.

Produto interno u·v=0,8; normas iguais a 1; similaridade 0,8. Para u·w=0; similaridade 0.

```python
import math
u, v, w = (1.0, 0.0), (0.8, 0.6), (0.0, 1.0)
def cosine(a, b):
    numerator = sum(x*y for x,y in zip(a,b))
    return numerator / (math.sqrt(sum(x*x for x in a)) * math.sqrt(sum(y*y for y in b)))
print("cos(u,v) =", cosine(u,v))
print("cos(u,w) =", cosine(u,w))
```
