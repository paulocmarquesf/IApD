## Exemplo mínimo verificável

Seis casos: antes acerta Q1,Q3,Q4,Q6; depois acerta Q1,Q2,Q4,Q5,Q6. A acurácia vai de 4/6 a 5/6; há duas melhorias (Q2,Q5) e uma regressão (Q3). Uma conclusão baseada somente no agregado esconde Q3.

Ganho líquido de 1 acerto; taxa inicial≈66,7%, final≈83,3%. A melhoria precisa ser interpretada em conjunto com a gravidade do caso regressivo.

```python
before = [True,False,True,True,False,True]
after = [True,True,False,True,True,True]
improvements = [i+1 for i,(a,b) in enumerate(zip(before,after)) if not a and b]
regressions = [i+1 for i,(a,b) in enumerate(zip(before,after)) if a and not b]
print("Antes:",sum(before),"Depois:",sum(after))
print("Melhorias:",improvements,"Regressões:",regressions)
```
