## Abrir os dados locais

A localização é identificada automaticamente. A demonstração utiliza arquivos fornecidos com o curso.

```python
from pathlib import Path
folder = Path.cwd()
for base in [folder, *folder.parents]:
    if (base / "dados").is_dir():
        DATA = base / "dados"
        break
else:
    raise FileNotFoundError("Abra o notebook a partir do repositório completo, que contém a pasta dados/")
print("Pasta de dados:", DATA)
```

## Exemplo mínimo verificável

O documento A afirma “risco externo elevado”. O documento B acrescenta “persistência de serviços”. A pergunta compara a ênfase nas duas reuniões. Recuperar apenas B permite identificar o tema, mas não sustenta a comparação temporal.

Para comparar duas reuniões, é preciso ao menos uma passagem pertinente de cada período. Um texto inventado que não aparece em nenhuma fonte não pode receber uma citação válida.

```python
from pathlib import Path
docs = [(p.name,p.read_text(encoding="utf-8")) for p in sorted((DATA / "copom").glob("*.md"))]
query = {"serviços", "inflação"}
ranked = sorted(docs,key=lambda x:sum(term in x[1].lower() for term in query),reverse=True)
for name,text in ranked[:2]:
    print(name, "\n", text[:200], "\n")
```
