## Exemplo mínimo verificável

Seis mensagens têm a referência Sim,Não,Sim,Não,Sim,Não e o sistema responde Sim,Não,Não,Não,Sim,Não. São cinco acertos: acurácia 5/6. O erro foi um falso negativo, pois um caso que exigia encaminhamento foi classificado como Não.

5÷6≈83,3%. Para encaminhamento humano, VP=2, VN=3, FN=1 e FP=0. A métrica agregada não mostra sozinha a consequência do FN.

```python
gold = [1,0,1,0,1,0]
pred = [1,0,0,0,1,0]
tp=sum(g==1 and p==1 for g,p in zip(gold,pred))
tn=sum(g==0 and p==0 for g,p in zip(gold,pred))
fp=sum(g==0 and p==1 for g,p in zip(gold,pred))
fn=sum(g==1 and p==0 for g,p in zip(gold,pred))
print({"accuracy":(tp+tn)/len(gold),"tp":tp,"tn":tn,"fp":fp,"fn":fn})
```
