## Exemplo mínimo verificável

Pedido entregue há cinco dias: consultar → verificar prazo → enviar para análise de elegibilidade. Pedido entregue há nove dias: consultar → verificar prazo → encaminhar para revisão humana. Sem número de pedido: solicitar complemento. Nenhum ramo executa reembolso automaticamente.

O caso de cinco dias apenas inicia análise de elegibilidade; a aprovação depende de outra ação. O grafo não deve equiparar elegibilidade a autorização.

```python
def route_request(days_since_delivery):
    if days_since_delivery is None:
        return ["receive", "request_id", "stop"]
    if days_since_delivery <= 7:
        return ["receive", "lookup_order", "check_policy", "eligibility_review", "stop"]
    return ["receive", "lookup_order", "check_policy", "human_review", "stop"]
for days in [5,9,None]:
    print(days,route_request(days))
```
