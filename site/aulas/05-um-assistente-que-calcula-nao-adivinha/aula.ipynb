## Exemplo mínimo verificável

Dois ativos: retornos esperados de 8% e 12%; volatilidades de 10% e 20%; correlação zero; pesos 50% e 50%. Retorno da carteira = 10%. Variância = (0,5²×0,10²)+(0,5²×0,20²)=0,0125. Volatilidade ≈11,18%.

A ferramenta de portfólio deve receber dois pesos, dois retornos, duas volatilidades e a correlação. Sem correlação, calcular a volatilidade com média simples de 10% e 20% seria incorreto.

```python
import math
w = 0.5
expected_return = w*0.08 + (1-w)*0.12
variance = w*w*0.10**2 + (1-w)**2*0.20**2
print({"expected_return":expected_return,"volatility":math.sqrt(variance),"sharpe":(expected_return-0.02)/math.sqrt(variance)})
```
