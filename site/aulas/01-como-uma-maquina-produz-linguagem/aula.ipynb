## Exemplo mínimo verificável

Para a continuação de “A entrega foi”, admita P(rápida)=0,60, P(lenta)=0,25 e P(cara)=0,15. Essas probabilidades somam 1. Se o contexto mudar para “Apesar de barata, a entrega foi”, a distribuição pode mudar. Isso não significa que o modelo verificou um fato; significa apenas que condicionou a previsão a um contexto diferente.

Se escolhermos sempre o token mais provável, nesta etapa a continuação será “rápida”. Se amostrarmos segundo a distribuição, “lenta” também pode aparecer. Precisão factual não equivale à probabilidade do próximo token.
