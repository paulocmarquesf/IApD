O documento didático do Delta informa 0,80% ao ano. A resposta pode usar esse número somente se recuperar o trecho correspondente, sem misturá-lo a outro custo ou produto.

```python
history = "Qual a taxa de administração do Fundo Horizonte?"
followup = "E a do Fundo Delta?"
attribute = "taxa de administração" if "taxa de administração" in history else "atributo não identificado"
rewritten = f"Qual é a {attribute} do Fundo Delta?"
print("Pergunta original:",followup)
print("Consulta reescrita:",rewritten)
```
