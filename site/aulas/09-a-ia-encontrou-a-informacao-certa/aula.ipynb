## Exemplo mínimo verificável

Quatro questões: A tem ranks 1,3,2,5; B tem 2,2,3,5. Para k=3, ambos têm Hit@3=3/4. MRR@3(A)=(1+1/3+1/2+0)/4≈0,458; MRR@3(B)=(1/2+1/2+1/3+0)/4≈0,333.

O documento na posição cinco não pontua em k=3. A ganha no MRR porque encontra o relevante na primeira posição em uma das perguntas.

```python
rank_a = [1,3,2,5]
rank_b = [2,2,3,5]
k = 3
def metrics(ranks,k):
    return (sum(r<=k for r in ranks)/len(ranks),sum(1/r if r<=k else 0 for r in ranks)/len(ranks))
print("A: Hit@3, MRR@3 =",metrics(rank_a,k))
print("B: Hit@3, MRR@3 =",metrics(rank_b,k))
```
